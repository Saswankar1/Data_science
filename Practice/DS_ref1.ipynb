## Numpy

- less computational time
- create n-d array

```python
import numpy as np
```

```python
#creating a numpy array
arr = np.array([1, 2, 3, 4, 5])
print(arr)
type(arr)
```

```python
arr2 = np.array([[1, 2, 3], [4, 5, 6]]) 
print(arr2)
type(arr2)
```

```python
arr3 = np.array([1, 2, 3], dtype=np.float64)
print(arr3)
type(arr3)
```

```python
arr2.ndim # this will return the number of dimensions of the array
```

```python
arr2.shape # this will return the shape of the array as a tuple (2, 3) which means 2 rows and 3 columns
```

```python
arr2[0, 1] # this will return the element at the first row and second column which is 2
```

```python
arr2[1:]
```

```python
print(np.zeros((2, 3))) # this will create a 2x3 array filled with zeros
print(np.ones((2, 3))) # this will create a 2x3 array filled with ones
print(np.full((2, 3), 7)) # this will create a 2x3 array filled with the value 7
```

```python
np.random.rand(2, 3) # this will create a 2x3 array filled with random values between 0 and 1
```

```python
np.random.randint(1, 10, size=(2, 3)) # this will create a 2x3 array filled with random integers between 1 and 10
```

```python
np.identity(2)
```

## Pandas

- data manipulation and open source data analysis

```python
import pandas as pd
```

```python
df = pd.read_csv('carprices.csv')
print(df.head())
```

```python
df.info()
```

```python
df.describe() #describe the data in the dataframe, it will give you the count, mean, std, min, 25%, 50%, 75% and max values for each column in the dataframe.
```

```python
df.columns # this will return the column names of the dataframe
```

```python
df["Sell Price($)"] += 100
df.head()
```

```python
df ["Discount"] = 10
df.head()
```

```python
df[4:7]
```

```python
df[df["Sell Price($)"] > 10000] # this will return all the rows where the sell price is greater than 10000
```

```python
df.pop("Discount") # this will remove the "Discount" column from the dataframe and return a new dataframe without the "Discount" column
df.head()
```
