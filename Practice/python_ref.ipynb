### Set Operations

```python
# Union()
set3 = {5, 6, 7, 8}
union_set = set2.union(set3)
union_set = set2 | set3 # this is another way to get the union of two sets
print(union_set)
```

```python
# intersection()
print(set2.intersection(set3)) # this will print the common elements between set2 and set3
print(set2 & set3) # this is another way to get the intersection of two sets    
```

```python
# difference()
print(set2.difference(set3)) # this will print the elements that are in set2 but not in set3
print(set2 - set3) # this is another way to get the difference of two sets
```

```python
# symmetric difference() -> this will print the elements that are in either set2 or set3 but not in both
print(set2.symmetric_difference(set3)) # this will print the elements that are in either set2 or set3 but not in both
print(set2 ^ set3) # this is another way to get the symmetric difference of two sets    
```

Sets are used to find the distinct count

```python
grades = ['A', 'A', 'B', 'C', 'A', 'B']
grades_set = set(grades) # this will convert the list of grades to a set of unique grades
print(grades_set)
```

## Funtions


```python
# lambda functions -> this is an anonymous function that can take any number of arguments but can only have one expression. The expression is evaluated and returned.
# syntax: lambda arguments: expression
x = 3
y = lambda x: x**2 # this will create a lambda function that takes one argument x and returns the square of x   
print(y(x))
```

```python
# factorial function
def factorial(n):
    fact = 1
    for i in range(1, n+1):
        fact *= i
    return fact 

factorial(5)
```

```python
# check if number is even or not using lambda function
is_even = lambda n: f"{n} is an even number" if n % 2 == 0 else f"{n} is an odd number"
print(is_even(4))
print(is_even(5))
```

```python
# Map, reduce, filter 

# map(function, iterable) -> this will apply the function to all the items in the iterable and return a map object (which is an iterator)
# map the list to obtain names in uppercase
names = ["Saswankar", "Raj", "Ankit"]
upper_names = map(lambda x: x.upper(), names) # this will apply the lambda function to all the items in the names list and return a map object
print(upper_names)
print(list(upper_names)) # this will convert the map object to a list and print the list of uppercase names

```

```python
# filter(function, iterable) -> this will construct an iterator from those elements of the iterable for which the function returns true. The function is called a predicate and is expected to return a boolean value.
# filter the list to obtain names that start with "S"
filtered_names = filter(lambda x: x.startswith("S"), names) 
print(list(filtered_names)) 
```

```python
# reduce(function, iterable) -> this will apply the function cumulatively to the items of the iterable from left to right, reducing the iterable to a single value. The function must take two arguments and return a single value. The reduce() function is not a built-in function in Python 3, but it is available in the functools module.
from functools import reduce    
reduce(lambda x, y: x + y, [1, 2, 3, 4]) # this will return the sum of all the elements in the list
```

```python
rect1 = Rectangle(10,20)
print(rect1.length)
print(rect1.width)
```

```python
# 2) Class variables and instance variables
class Circle:
    pi = 3.14 # this is a class variable because it is shared by all instances of the class
    def __init__(self, radius):
        self.radius = radius # this is an instance variable because it is unique to each instance of the class
```

```python
circle1 = Circle(5)
circle2 = Circle(10)
print("Radius of circle1:", circle1.radius)
print("Pi of circle1:", circle1.pi)
print("Radius of circle2:", circle2.radius)
print("Pi of circle2:", circle2.pi)
```

```python
circle1.pi = 3.14159 # this will create a new instance variable pi for circle1 and will not modify the class variable pi
print("Pi of circle1:", circle1.pi) # this will print the instance variable pi
print("Pi of circle2:", circle2.pi) # this will print the class variable pi
```

```python
class Square:
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2
```

```python
sq = Square(4)
print(sq.area())
```

```python
# Inheritance -> this is a way to create a new class that is a modified version of an existing class. The new class is called the child class and the existing class is called the parent class. The child class inherits all the attributes and methods of the parent class and can also have its own attributes and methods.
class Employee:
    def function(self):
        print("This is a function in the Employee class")

class Manager(Employee): # this will create a new class Manager that inherits from the Employee class
    pass
```

```python
emp = Employee()
emp.function()
```

```python
dept = Manager()
dept.function() # this will call the function() method of the Employee class because the Manager class inherits from the Employee class and does not have its own function() method.
```

```python
# Overriding -> this is a way to modify the behavior of a method in the child class that is already defined in the parent class. The child class can provide
class Shape:
    def set_color(self, color):
        self.color = color
    def area(self):
        pass
    def color_the_shape(self):
        color_price = {"red": 10, "blue": 20, "green": 30}
        return self.area() * color_price[self.color]
```

```python
class circle(Shape):
    pi = 3.14
    def __init__(self, radius):
        self.radius = radius
    def area(self): # this will override the area() method of the Shape class   
        return self.pi * self.radius ** 2
```

```python
c = circle(5)
c.set_color("red")
print("Circle with radius", c.radius, "when colored using", c.color, "has a cost of", c.color_the_shape())
```
